```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

symmetric = np.random.normal(50, 10, 1000)
right_skewed = np.random.exponential(10, 1000)
left_skewed = -np.random.exponential(10, 1000)

fig, ax = plt.subplots(1, 3, figsize=(12, 4))

ax[0].hist(symmetric, bins=30)
ax[0].set_title("Approximately Symmetric")

ax[1].hist(right_skewed, bins=30)
ax[1].set_title("Right-Skewed")

ax[2].hist(left_skewed, bins=30)
ax[2].set_title("Left-Skewed")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)

normal_data = np.random.normal(50, 10, 1000)
skewed_data = np.random.exponential(10, 1000)

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].hist(normal_data, bins=30)
ax[0].set_title("Normal Distribution")

ax[1].hist(skewed_data, bins=30)
ax[1].set_title("Right-Skewed Distribution")

plt.tight_layout()
plt.show()
```
