```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Headphones"],
    "Sales": [45, 80, 30, 60]
})

df.plot(x = "Product", y = "Sales", kind = "bar")

plt.title("Product Sales")
plt.ylabel("Units sold")
plt.xlabel("Product")
plt.xticks(rotation=45)
plt.show()
```
