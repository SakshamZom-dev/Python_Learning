```python
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')      # or (1,1,1) could also be used in place of (111)

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]
z = [5, 3, 6, 2, 7]

ax.scatter(x, y, z, color='red')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
```
