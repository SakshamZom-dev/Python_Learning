```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 2.5, 3, 4, 5, 6, 7]
marks = [35, 42, 48, 50, 62, 70, 78, 88]

plt.scatter(study_hours, marks, color="purple", alpha=0.8)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.grid(True, alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 15000, 13500, 18000, 21000, 25000]

plt.plot(
    months,
    revenue,
    marker="o",
    color="green",
    linewidth=2
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.grid(True, alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ["Phones", "Laptops", "Tablets", "Accessories"]
sales = [120, 95, 60, 80]

plt.bar(categories, sales, color="teal")

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Units Sold")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

categories = ["Phones", "Laptops", "Tablets"]
sales_2025 = [100, 80, 50]
sales_2026 = [120, 95, 65]

x = np.arange(len(categories))
width = 0.35

plt.bar(x - width / 2, sales_2025, width, label="2025")
plt.bar(x + width / 2, sales_2026, width, label="2026")

plt.xticks(x, categories)
plt.ylabel("Units Sold")
plt.title("Product Sales: 2025 vs 2026")
plt.legend()
plt.show()
```

```python

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 90, 150, 170, 160]
profit = [20, 25, 12, 30, 38, 35]

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Left: Monthly sales trend
ax[0].plot(months, sales, marker="o", color="blue")
ax[0].set_title("Monthly Sales")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales (Units)")
ax[0].grid(True, alpha=0.3)

# Right: Relationship between sales and profit
ax[1].scatter(sales, profit, marker="o", color="green")
ax[1].set_title("Sales vs Profit")
ax[1].set_xlabel("Sales (Units)")
ax[1].set_ylabel("Profit (₹)")
ax[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
