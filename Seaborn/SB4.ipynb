```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.barplot(data=data, x='day', y='total_bill', color= 'lime', alpha=0.5)
plt.title('Average Total Bill by Day')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.barplot(data=data, x='day', y='total_bill', estimator=sum, color= 'yellow', alpha=0.5)
plt.title('Total Bill by Day')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.barplot(data=data, x='day', y='total_bill', errorbar=None, color= 'skyblue', alpha=0.5)
plt.title('Average Total Bill by Day')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.barplot(data=data, x='day', y='total_bill', hue='sex', palette=['blue', 'violet'], alpha=0.5)
plt.title('Total Bill by Day and Sex')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.barplot(data=data, x='day', y='tip', hue='sex', palette='Set2')
plt.title('Average Tip by Day and Sex')
plt.show()
```
