```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")
sns.histplot(data = data, x = "total_bill", color="aqua")

plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data=data, x='total_bill', ax=ax[0])
ax[0].set_title("Distribution of Total Bill")
ax[0].set_xlabel("Total Bill")

sns.histplot(data=data, x='tip', ax=ax[1])
ax[1].set_title("Distribution of Tip")
ax[1].set_xlabel("Tip")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

data = sns.load_dataset('tips')

sns.displot(data=data, x='total_bill')    # displot for total_bill
sns.displot(data=data, x='tip', kde=True)    # displot for tip

plt.show()
    
```
