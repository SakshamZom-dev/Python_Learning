```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.pairplot(data[['total_bill', 'tip', 'size']])

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.pairplot(
    data,
    vars=['total_bill', 'tip', 'size'],
    hue='sex',
    diag_kind='kde',
    corner=True
)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.pairplot(
    data,
    vars=['total_bill', 'tip', 'size'],
    hue='smoker',
    kind='scatter',
    diag_kind='hist',
    palette=['violet', 'maroon'],
    corner=True
)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.jointplot(
    data= data,
    x='total_bill',
    y= 'tip',
    kind='scatter',
)

plt.show()
```
