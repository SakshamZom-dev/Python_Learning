```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.countplot(data=data, x='day', color='orange', alpha=0.5)

plt.title('Customers by Day')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.countplot(data=data, x='day', hue='sex', palette=['navy', 'magenta'], alpha=0.5)

plt.title('Customers by Day and Sex')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.countplot(data=data, x='day', order=['Fri','Thur', 'Sat', 'Sun'], hue='sex', palette=['violet', 'purple'], alpha=0.5)

plt.title('Customers by Day and Sex')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.countplot(data=data, x='time', hue='smoker', palette=['lightblue', 'salmon'], alpha=0.7)

plt.title('Customers by Meal Time')
plt.show()
```
