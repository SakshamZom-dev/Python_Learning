```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.histplot(data= data, x= 'total_bill', bins= 20, color = 'skyblue', kde=True)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.kdeplot(data= data, x= 'total_bill', fill= True, color= 'orange')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.kdeplot(data= data, x= 'total_bill', bw_adjust= 2)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.kdeplot(data= data, x= 'total_bill', fill= True, bw_adjust= 2, hue= 'sex')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

sns.kdeplot(data = data, x= 'total_bill', fill= True, color= 'purple')
plt.title('Restaurant Bill Distribution')
plt.xlabel('Total Bill')

plt.show()
```
