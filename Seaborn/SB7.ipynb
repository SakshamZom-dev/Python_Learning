```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

correlation = data[['total_bill', 'tip', 'size']].corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap of Tips Dataset')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

correlation = data[['total_bill', 'tip', 'size']].corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm', linewidths=0.5, fmt='.2f', vmin=-1, vmax=1)

plt.title('Correlation Heatmap of Tips Dataset')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

correlation = data[['total_bill', 'tip', 'size']].corr()

sns.heatmap(correlation, annot=True, cmap='viridis', linewidths=0.5)

plt.title('Tips Dataset Correlation')
plt.show()
```
