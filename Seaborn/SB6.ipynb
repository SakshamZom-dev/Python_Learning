```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")
sns.scatterplot(data=data, x="total_bill", y="tip")
plt.title("Total Bill vs Tip")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.scatterplot(
    data=data,
    x='total_bill',
    y='tip',
    hue='sex',
    style='smoker',
    size='size'
)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.scatterplot(
    data=data,
    x='total_bill',
    y='tip',
    hue='day',
    alpha=0.7,
    size='size'
)

plt.title("Bill vs Tip Analysis")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.lineplot(data= data, x="size", y="total_bill")
plt.title("Party Size vs Total Bill")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.lineplot(data= data, x="size", y="total_bill", errorbar=None, estimator='sum')
plt.title("Party Size vs Total Bill")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.lineplot(data= data, x="size", y="total_bill", hue="sex", errorbar=None, marker='o')
plt.title("Party Size vs Total Bill by Sex")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.relplot(data= data, x="total_bill", y="tip", hue="sex", col="time", kind="scatter")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.relplot(data= data, x="size", y="total_bill", hue="smoker", col='time', kind="scatter")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

g = sns.relplot(
    data=data,
    x='size',
    y='total_bill',
    hue='smoker',
    col='time',
    kind='scatter'
)

g.fig.suptitle("Party Size vs Total Bill by Smoker Status and Time")
g.fig.subplots_adjust(top=0.85)

plt.show()
```
