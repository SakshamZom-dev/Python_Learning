```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.regplot(data= data, x="total_bill", y="tip", scatter= True, fit_reg= True, ci= 95, color= "purple", marker= "o", line_kws={"color": "red", "linewidth": 2})
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

sns.lmplot(
    data= data, 
    x="total_bill", 
    y="tip",
    hue= "sex",
    row= "smoker",
    height= 4,
    aspect= 1.2
)

plt.show()
```
