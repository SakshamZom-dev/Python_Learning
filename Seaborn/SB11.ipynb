```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# First plot
sns.histplot(data=data, x="total_bill", kde=True, ax=ax[0])
ax[0].set_title("Bill Distribution")
ax[0].set_xlabel("Total Bill")
ax[0].set_ylabel("Frequency")

# Second plot
sns.scatterplot(data=data, x="total_bill", y="tip", ax=ax[1])
ax[1].set_title("Bill vs Tip")
ax[1].set_xlabel("Total Bill")
ax[1].set_ylabel("Tip")

fig.suptitle("Restaurant Data Analysis", fontsize=16)

plt.tight_layout()
plt.show()
```
