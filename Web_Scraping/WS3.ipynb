```python
import requests
from bs4 import BeautifulSoup

from urllib.parse import urljoin


base_url = "https://books.toscrape.com/"

response = requests.get(base_url)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

books = soup.select("article.product_pod")

for book in books:

    title = book.h3.a["title"]

    price = book.select_one(".price_color").get_text(strip=True)

    rating = book.select_one(".star-rating")["class"][1]

    book_link = urljoin(base_url, book.h3.a["href"])            # Converting relative URLs into absolute URLs

    image_url = urljoin(base_url, book.img["src"])              # Convert the image path into an absolute URL

    print("Title:", title)
    print("Price:", price)
    print("Rating:", rating)
    print("Book URL:", book_link)
    print("Image URL:", image_url)
    print("-" *60)

print("\nTotal books scraped:", len(books))
```

```python
import requests
from bs4 import BeautifulSoup

# 1. Fetch the webpage
url = "https://www.scrapethissite.com/pages/forms/"

response = requests.get(url)
response.raise_for_status()

# 2. Parse the HTML
soup = BeautifulSoup(response.text, "html.parser")

# 3. Find the table
table = soup.find("table")

# 4. Get all table rows
rows = table.find_all("tr")

# 5. Extract column headings
cells = rows[0].find_all(["th", "td"])

headers = []

for cell in cells:
    headers.append(cell.get_text(strip=True))

# 6. Extract data from remaining rows
data = []

for row in rows[1:]:
    data_cells = row.find_all("td")

    values = []

    for cell in data_cells:
        values.append(cell.get_text(strip=True))

    if values:                                  # This checks whether the values list contains anything.
        record = dict(zip(headers, values))     # zip() pairs corresponding elements & dict() converts those pairs into a dictionary
        data.append(record)

# 7. Display structured data
for record in data:
    print(record)

print("\nTotal records:", len(data))
```
